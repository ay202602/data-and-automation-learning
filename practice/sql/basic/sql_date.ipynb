```python
# 日付関数クエリ練習用ファイル
import pandas as pd
import sqlite3
```

```python
class Sql:
    def __init__(self, df:pd.DataFrame, conn:sqlite3.Connection):
        self.df = df
        self._conn = conn
    
    
    def conversion_sql(self, name:str,) -> None:
        data = self.df.to_sql(
            name=name,
            con=self._conn,
            if_exists="replace",
            index=False
        )
        return data
    
    
    def result_sql(self, query:str) -> None:
        result = pd.read_sql(
            query,
            con=self._conn
        )
        return print(result)
```

```python
df = pd.DataFrame({
    "id": [1, 2, 3, 4, 5, 6],
    "name": ["Aさん", "Bさん", "Cさん", "Dさん", "Eさん", "Fさん"],
    "department": ["経理", "経理", "開発", "総務", "総務", "開発"],
    "salary": [1200, 1500, 10000, 2000, 1087, 1100],
    "start_date": pd.to_datetime(
        ["2023-4-1", "2023-4-1", "2000-12-30", "2023-4-1", "2000-12-30", "2020-6-12"]
    )    
})
print(df)
```

```python
conn = sqlite3.connect(":memory:")
sql = Sql(df=df, conn=conn)
sql.conversion_sql("start")
```

```python
# 日付から年月を取り出す
sql.result_sql(
    """SELECT department, strftime('%Y-%m', start_date) AS year_month, SUM(salary) AS total_salary
        FROM start
        GROUP BY department, strftime('%Y-%m', start_date)
    """
)
```

```python
# DataFrame内の日付差（日数）を表示
sql.result_sql(
    """SELECT department, CAST(julianday('now') - julianday(start_date) AS INTEGER) AS day_start
        FROM start
    """
)
```
